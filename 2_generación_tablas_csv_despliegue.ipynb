```python
query_1 = """
select 
    estudiantes.id_estudiante,
    estudiantes.nombre,
    estudiantes.email,
    bootcamp.nombre as bootcamp,
    campus.ubicacion as sede,
    promociones.nombre as promocion,
    promociones.fecha_comienzo
from estudiantes
join edicion on estudiantes.id_edicion = edicion.id_edicion
join bootcamp on edicion.id_bootcamp = bootcamp.id_bootcamp
join campus on edicion.id_campus = campus.id_campus
join promociones on edicion.id_promocion = promociones.id_promocion
order by estudiantes.id_estudiante;
"""

query_2 = """
select 
    campus.ubicacion as sede,
    bootcamp.nombre as bootcamp,
    count(estudiantes.id_estudiante) as total_estudiantes
from estudiantes
join edicion on estudiantes.id_edicion = edicion.id_edicion
join campus on edicion.id_campus = campus.id_campus
join bootcamp on edicion.id_bootcamp = bootcamp.id_bootcamp
group by campus.ubicacion, bootcamp.nombre
order by campus.ubicacion, bootcamp.nombre;
"""

query_3 = """
select 
    bootcamp.nombre as bootcamp,
    proyectos.nombre as proyecto,
    count(*) as total_proyectos,
    sum(case when calificaciones.calificacion = 'Apto' then 1 else 0 end) as proyectos_aprobados,
    round(
        100.0 * sum(case when calificaciones.calificacion = 'Apto' then 1 else 0 end) / count(*), 
        2
    ) as porcentaje_aprobados
from calificaciones
join proyectos on calificaciones.id_proyecto = proyectos.id_proyecto
join bootcamp on proyectos.id_bootcamp = bootcamp.id_bootcamp
group by bootcamp.nombre, proyectos.id_proyecto, proyectos.nombre
order by bootcamp.nombre, proyectos.id_proyecto;
"""

query_4 = """
select 
    profesores.nombre as profesor,
    profesores.rol,
    bootcamp.nombre as bootcamp,
    campus.ubicacion as sede,
    modalidades.nombre as modalidad,
    promociones.nombre as promocion
from asignacion_claustro
join profesores on asignacion_claustro.id_profesor = profesores.id_profesor
join bootcamp on asignacion_claustro.id_bootcamp = bootcamp.id_bootcamp
join campus on asignacion_claustro.id_campus = campus.id_campus
join modalidades on asignacion_claustro.id_modalidad = modalidades.id_modalidad
join promociones on asignacion_claustro.id_promocion = promociones.id_promocion
order by campus.ubicacion, bootcamp.nombre, profesores.rol;
"""

query_5 = """
select 
    estudiantes.nombre as estudiante,
    bootcamp.nombre as bootcamp,
    campus.ubicacion as sede,
    count(calificaciones.id_proyecto) as calificacion
from estudiantes
join edicion on estudiantes.id_edicion = edicion.id_edicion
join bootcamp on edicion.id_bootcamp = bootcamp.id_bootcamp
join campus on edicion.id_campus = campus.id_campus
join calificaciones on estudiantes.id_estudiante = calificaciones.id_estudiante
where calificaciones.calificacion = 'Apto'
group by estudiantes.id_estudiante, estudiantes.nombre, bootcamp.nombre, campus.ubicacion
having count(calificaciones.id_proyecto) = 5
order by bootcamp.nombre, estudiantes.nombre;
"""
```
