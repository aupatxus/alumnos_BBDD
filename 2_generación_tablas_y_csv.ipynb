```python
# Creamos el diccionario {nombre_proyecto: posicion} usando enumerate
mapa_proyectos = {proyecto: i for i, proyecto in enumerate(lista_proyectos)}
print(mapa_proyectos)
```

```python
df_proyectos_estudiantes = df_clases[lista_proyectos]
```

```python
df_proyectos_estudiantes.head()
```

```python
df_proyectos_estudiantes = df_proyectos_estudiantes.melt()
```

```python
df_proyectos_estudiantes
```

```python
df_proyectos_estudiantes.info()
```

```python
list_notas = []

for indice, fila in df_clases.iterrows():
    id_estudiante = indice
    for proyecto in lista_proyectos:
        calificacion = fila[proyecto]  
        if pd.notna(calificacion):
            id_proyecto = mapa_proyectos[proyecto] 
            list_notas.append({
                "id_estudiante": id_estudiante,
                "id_proyecto": id_proyecto,
                "calificacion": calificacion
            })

# Convertimos la lista de diccionarios en DataFrame
df_calificaciones = pd.DataFrame(list_notas)
df_calificaciones = df_calificaciones.set_index("id_estudiante")

df_calificaciones.head()
```
